```python
# week2 excercise
# Apply the X gate (also known as a NOT gate) to change the qubit to the ∣1⟩
# state.
# • Visualize the ∣1⟩ state on the Bloch sphere.
from qiskit.visualization import plot_bloch_vector
import qiskit as qk
from qiskit.quantum_info import Statevector, Pauli

qc_one = qk.QuantumCircuit(1)
qc_one.x(0)

state = Statevector(qc_one)
plot_bloch_multivector(state)
```

```python
# week2 excercise
#  Apply the Hadamard gate to create a superposition state.
# • Visualize how the state changes on the Bloch sphere.
from qiskit.visualization import plot_bloch_multivector
import qiskit as qk
from qiskit.quantum_info import Statevector

qc_one = qk.QuantumCircuit(2)
qc_one.h(1)

state = Statevector(qc_one)
plot_bloch_multivector(state)
```
